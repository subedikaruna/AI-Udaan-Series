```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dd = pd.read_csv("datasets/dp_lms_student_analytics_dataset.csv")
dd["EnrollmentDate"] = pd.to_datetime(dd["EnrollmentDate"])
dd["Month"] = dd["EnrollmentDate"].dt.strftime("%b")

dd.columns
# dd.head()
```

```python

monthly_enrollment = dd.groupby("Month")["StudentID"].count()
plt.figure(figsize=(10,5))
plt.plot(monthly_enrollment.index, monthly_enrollment.values,linewidth=2)
plt.title("Monthly Student Enrollment")
plt.xlabel("Month")
plt.ylabel("Number of Students")
plt.grid(True)
plt.show()
```

```python
monthly_revenue = dd.groupby("Month")["Revenue"].sum()

plt.figure(figsize=(10,5))

plt.plot(monthly_revenue.index,
         monthly_revenue.values,
         linewidth=2)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.grid(True)

plt.show()
```

```python
monthly_progress = dd.groupby("Month")["CourseProgress"].mean()

plt.figure(figsize=(10,5))

plt.plot(monthly_progress.index,
         monthly_progress.values,
         color='purple',
         linewidth=2)

plt.title("Average Course Progress by Month")
plt.xlabel("Month")
plt.ylabel("Average Progress (%)")

plt.grid(True)

plt.show()
```

```python
monthly_attendance = dd.groupby("Month")["AttendancePercentage"].mean()

plt.figure(figsize=(10,6))

plt.plot(monthly_attendance.index,
         monthly_attendance.values,

         color='pink',
         linewidth=2)

plt.title("Average Attendance Percentage by Month")
plt.xlabel("Month")
plt.ylabel("Attendance (%)")

plt.grid(True)

plt.show()
```

```python
course_students = dd.groupby("Course")["StudentID"].count()

plt.figure(figsize=(12,8))

plt.bar(course_students.index,
        course_students.values,
        color="blue")

plt.title("Students in Each Course")
plt.xlabel("Course")
plt.ylabel("Students")

plt.xticks(rotation=45)

plt.grid()

plt.show()
```

```python
course_revenue = dd.groupby("Course")["Revenue"].sum()

plt.figure(figsize=(12,8))

plt.bar(course_revenue.index,
        course_revenue.values,
        color="pink")

plt.title("Revenue Generated by Each Course")
plt.xlabel("Course")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.grid()


plt.show()
```

```python
course_score = dd.groupby("Course")["FinalScore"].mean()

plt.figure(figsize=(12,8))

plt.bar(course_score.index,
        course_score.values,
        color="darkgreen")

plt.title("Final Score by Course")
plt.xlabel("Course")
plt.ylabel("Average Final Score")

plt.xticks(rotation=30)

plt.grid()

plt.show()
```

```python
course_attendance = dd.groupby("Course")["AttendancePercentage"].mean()

plt.figure(figsize=(12,8))

plt.bar(course_attendance.index,
        course_attendance.values,
        color="grey")

plt.title("Attendance % by Course")
plt.xlabel("Course")
plt.ylabel("Attendance Percentage")

plt.xticks(rotation=25)

plt.grid()


plt.show()
```

```python
city_students = dd.groupby("City")["StudentID"].count()

plt.figure(figsize=(12,8))

plt.bar(city_students.index,
        city_students.values,
        color="purple")

plt.title("Students from Each City")
plt.xlabel("City")
plt.ylabel("Students")
plt.xticks(rotation=25)
plt.grid()
plt.show()
```

```python
instructor_students = dd.groupby("Instructor")["StudentID"].count()

plt.figure(figsize=(12,8))

plt.bar(instructor_students.index,
        instructor_students.values,
        color="chocolate")
plt.title("Number of Students Under Each Instructor")
plt.xlabel("Instructor")
plt.ylabel("Students")

plt.grid()
plt.show()
```

```python
top_courses = dd.groupby("Course")["StudentID"].count().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,8))
plt.barh(top_courses.index, top_courses.values, color="lightpink")

plt.title("Top 10 Courses Enrolled")
plt.xlabel("Students")
plt.ylabel("Course")
plt.grid()
plt.show()
```

```python
top_instructors = dd.groupby("Instructor")["StudentID"].count().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,8))
plt.barh(top_instructors.index, top_instructors.values, color="yellow")
plt.title("Top 10 Instructors by Student Enrollment")
plt.xlabel("Students")
plt.ylabel("Instructor")
plt.grid()
plt.show()
```

```python
top_students = dd.sort_values("FinalScore", ascending=False).head(10)
plt.figure(figsize=(10,8))
plt.barh(top_students["StudentName"], top_students["FinalScore"], color="lightblue")
plt.title("Top 10 Students by Final Score")
plt.xlabel("Final Score")
plt.ylabel("Student")
plt.grid()
plt.show()
```
