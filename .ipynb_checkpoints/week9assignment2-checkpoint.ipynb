```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    classification_report
)
```

```python
df = pd.read_csv(
    "datasets/student_multiclass_multiple_features.csv"
)

print("First 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)


X = df[
    [
        "StudyHours",
        "Attendance",
        "PreviousMarks"
    ]
]



y = pd.DataFrame({

    "MathSupport":
        (df["PreviousMarks"] < 50).astype(int),

    "EnglishSupport":
        (df["StudyHours"] < 4).astype(int),

    "AttendanceSupport":
        (df["Attendance"] < 75).astype(int)

})


print("\nTarget labels:")
print(y)


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)



scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)



model = OneVsRestClassifier(
    LogisticRegression(
        max_iter=1000
    )
)



model.fit(
    X_train_scaled,
    y_train
)


y_pred = model.predict(
    X_test_scaled
)


print("\nPredictions:")
print(y_pred)



print("\n================================")
print("MULTILABEL MODEL PERFORMANCE")
print("================================")

print(
    "Subset Accuracy:",
    accuracy_score(
        y_test,
        y_pred
    )
)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "Math Support",
            "English Support",
            "Attendance Support"
        ],
        zero_division=0
    )
)



new_student = pd.DataFrame({

    "StudyHours": [5.5],

    "Attendance": [85],

    "PreviousMarks": [65]

})


new_student_scaled = scaler.transform(
    new_student
)


prediction = model.predict(
    new_student_scaled
)[0]



print("\n================================")
print("STUDENT SUPPORT PREDICTION")
print("================================")

print(
    "Math Support:",
    "YES" if prediction[0] == 1 else "NO"
)

print(
    "English Support:",
    "YES" if prediction[1] == 1 else "NO"
)

print(
    "Attendance Support:",
    "YES" if prediction[2] == 1 else "NO"
)
```
