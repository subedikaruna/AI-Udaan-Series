```python
dd.duplicated().sum()
dd = dd.drop_duplicates()
```

```python
dd.isnull().sum()
num_cols = dd.select_dtypes(include=["int64","float64"]).columns
for col in num_cols:
    dd[col] = dd[col].fillna(dd[col].mean())
    text_cols = dd.select_dtypes(include=["object", "string"]).columns

for col in text_cols:
    dd[col] = dd[col].fillna(dd[col].mode()[0])


```

```python
dd["enrollment_date"] = pd.to_datetime(dd["enrollment_date"], errors="coerce")


numeric_columns = [
    "course_fee",
    "discount",
    "attendance_percent",
    "assignment_score",
    "final_score",
    "rating",
    "age"
]

for col in numeric_columns:
    dd[col] = pd.to_numeric(dd[col], errors="coerce")
    
for col in text_cols:
    dd[col] = (
        dd[col]
        .astype(str)
        .str.strip()
        .str.title()
    ) 

negative_columns = [
    "course_fee",
    "attendance_percent",
    "assignment_score",
    "final_score",
    "age",
    "discount",
    "rating"
]

for col in negative_columns:
    dd[col] = dd[col].astype(float)

    mean_value = dd.loc[dd[col] >= 0, col].mean()

    dd.loc[dd[col] < 0, col] = mean_value
    

```

```python
columns = [
    "course_fee",
    "attendance_percent",
    "assignment_score",
    "final_score"
]

outliers = {}

for col in columns:

    Q1 = dd[col].quantile(0.25)
    Q3 = dd[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

 


    outlier_rows = dd[(dd[col] < lower) | (dd[col] > upper)]

    outliers[col] = (lower, upper)



dd_remove = dd.copy()

for col in columns:

    lower, upper = outliers[col]

    dd_remove = dd_remove[
        (dd_remove[col] >= lower) &
        (dd_remove[col] <= upper)
    ]


print(dd_remove.shape)



dd_median = dd.copy()

for col in columns:

    lower, upper = outliers[col]

    median = dd[col].median()

    dd_median.loc[
        (dd_median[col] < lower) |
        (dd_median[col] > upper),
        col
    ] = median





dd_clip = dd.copy()

for col in columns:

    lower, upper = outliers[col]

    dd_clip[col] = dd_clip[col].clip(lower, upper)



z_columns = [
    "course_fee",
    "attendance_percent",
    "assignment_score",
    "final_score"
]

for col in z_columns:

    mean = dd[col].mean()
    std = dd[col].std()

    dd[col + "_zscore"] = (dd[col] - mean) / std

    outliers = dd[(dd[col + "_zscore"] > 3) | (dd[col + "_zscore"] < -3)]
outliers
   
```
