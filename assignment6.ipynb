```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("datasets/student_marks.csv")
df
df.head(5)
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
```

```python
X=df[["StudyHours"]]#features(input)
Y=df["FinalMarks"]#label

plt.scatter(X,Y)
plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("Study Hours Vs Final Marks")
X=X.to_numpy()
Y=Y.to_numpy()
#machine learning
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(X_train,Y_train)
predicted_Y=model.predict(X_test)

comparison=pd.DataFrame({
    "Actual":Y_test,
    "Predicted":predicted_Y
})
comparison
# new_students=np.array([
#             [2],
#             [4],
#             [6],
#             [8],
#             [10],
# ])
# marks_prediction=model.predict(new_students)
# marks_prediction
```

```python
df=pd.read_csv("datasets/student_multiple_features.csv")
print(df.head(5))
print(df.tail(5))
print(df.shape)
print(df.info())
print(df.describe())
print(df.isna().sum())
df["Attendance"]=pd.to_numeric(df["Attendance"],errors="coerce")
df.isnull().sum()
df["Attendance"]=df["Attendance"].fillna(df["Attendance"].median())
df["PreviousMarks"]=df["PreviousMarks"].fillna(df["PreviousMarks"].median())
df

#machine learning (feature segmentation)

X=df[
    [
        "StudyHours",
        "Attendance",
        "PreviousMarks"
    ]
    ]
Y=df["FinalMarks"]
X=X.to_numpy()
Y=Y.to_numpy()
#split (training,testing)
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(X_train,Y_train)
new_students=np.array([
   [10,80,60]
])
prediction=model.predict(new_students)
prediction


print("Intercept (Bias):", model.intercept_)

print("Coefficient (Weight) of StudyHours:", model.coef_[0])

print("Coefficient (Weight) of Attendance:", model.coef_[1])

print("Coefficient (Weight) of PreviousMarks:", model.coef_[2])


new_students = np.array([
    [5, 90, 70],   # Student A
    [7, 95, 82],   # Student B
    [3, 75, 55]    # Student C
])

predictions = model.predict(new_students)

print("\nPredicted Final Marks:")

print("Student A:", predictions[0])
print("Student B:", predictions[1])
print("Student C:", predictions[2])
```
