```python
sd=pd.DataFrame({
    "Salary":[-81911,1000,2000,3000,5000,7000,8000,9000,12000,90000]
})
#quartiles
median=sd["Salary"].quantile(0.5)
# median
q1=sd["Salary"].quantile(0.25)

q3=sd["Salary"].quantile(0.75)
# q1
# q3
iqr=q3-q1
# iqr
lower_bound=q1-(1.5*iqr)
# lower_bound
upper_bound=q3+(1.5*iqr)
upper_bound
```

```python
outliers=sd[(sd["Salary"]<lower_bound) | (sd["Salary"]>upper_bound)]
print(outliers)
```
