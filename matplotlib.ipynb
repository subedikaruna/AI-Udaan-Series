```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
index=[1,2,3,4,5]#x axis ma dekhauney
price=[1000,2000,3000,4000,4500]#y-axis ma dekhauney
# plt.figure(figsize=(16,2))#default if this line is not given then figsize is(6,6)
plt.plot(index,price)
# plt.bar(index,price)
```

```python
dd=pd.read_csv("datasets/daraz_dataset.csv")
dd
```

```python
category=dd.groupby("Category")["Product"].count()
# category
# category.index
# category.values
plt.figure(figsize=(18,4))
plt.bar(category.index,category.values,color="green",edgecolor="black")
plt.xlabel("Categories",fontsize=20)
plt.ylabel("Products",fontsize=20)
plt.title("Categories & Products Bar Graph",fontsize=20)
plt.grid(axis="y",linestyle="--",alpha=1)#aplha defines transparency of linestyle
```

```python
plt.figure(figsize=(18,6))
colors=["green","brown","pink","yellow","blue","red","black"]
plt.pie(category.values,labels=category.index,colors=colors,shadow=True,startangle=90,explode=(0.1,0,0,0,0,0.1,0))
plt.title("Product Distribution By Category")
```

```python
gender=dd.groupby("Gender")["Product"].count()
plt.figure(figsize=(18,10))
plt.pie(gender.values,labels=gender.index)
plt.title("Product By Gender")
```

```python
bins=[0,5000,10000,20000,50000,80000]
plt.hist(dd["UnitPrice"],bins=bins,color="green",edgecolor="black",alpha=0.5)
plt.title("Distribution Of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
```

```python
data={
    "Age":[20,22,24,26,28,30],
    "TotalAmount":[1000,1200,1500,2000,3000,32000]
    
}
dfd=pd.DataFrame(data)
plt.scatter(dfd["Age"],dfd["TotalAmount"],s=80)#marker size =s
plt.xlabel("Age")
plt.ylabel("Total Amount")
plt.grid(True)
plt.title("Age vs Total Amount")
```
